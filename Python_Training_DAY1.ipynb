```python
# Accept 5 paper marks, rollno, name;and calculate total and percentage.if percentage >=85, student can take addmission else can't take admission, using class & object & constant


```

```python

class Student:
    MAX_MARKS = 500  # Constant

    def accept(self):
        self.rollno = int(input("Enter Roll Number: "))
        self.name = input("Enter Student Name: ")

        self.marks = []
        print("Enter marks of 5 subjects:")

        for i in range(5):
            mark = float(input("Enter marks: "))
            self.marks.append(mark)

    def calculate(self):
        self.total = sum(self.marks)
        self.percentage = (self.total / Student.MAX_MARKS) * 100

    def display(self):
        print("\n----- Student Details -----")
        print("Roll Number:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", self.total)
        print("Percentage:", self.percentage, "%")

        if self.percentage >= 85:
            print("Admission: Student can take admission")
        else:
            print("Admission: Student cannot take admission")


# Creating Object
s1 = Student()

s1.accept()
s1.calculate()
s1.display()

```

```python
#Single Inheritance: A type of inheritance where a single child class inherits properties and methods from a single parent class.
#Multiple Inheritance: A type of inheritance where a single child class inherits from more than one parent class (supported in Python, but not in languages like Java).
#Hierarchical Inheritance: A type of inheritance where multiple child classes inherit from a single parent class.
#Hybrid Inheritance: A combination of two or more types of inheritance (for example, combining Multilevel and Multiple inheritance).
```

```python
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C(B, A): # Changed (A, B) to (B, A) to respect MRO
  def showC(self):
    print("showC function")

obj = C()
obj.showA()
obj.showB()
obj.showC()
```

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

print(" Object B ")
obj1 = B()
obj1.showA()
obj1.showB()
print("Object C")
obj2 = C()
obj2.showA()
obj2.showC()
```

```python
#hybrid inheritance

class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

class D(B, C):  # Inherits from both B and C
    def showD(self):
        print("showD function")
```
