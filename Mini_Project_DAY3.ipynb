```python
import pandas as pd
data = {
    'Name': ['A','B','C','D','E'],
    'Maths': [80,27,30,58,98],
    'Scince': [90,24,25,65,18],
    'English': [50,28,35,55,48]
}
df = pd.DataFrame(data)
print(df)
```

```python
import pandas as pd

# Corrected 'Scince' to match the actual DataFrame column spelling
marks = df[["Maths", "Scince", "English"]].values
average_marks = marks.mean(axis=1)
df["Average Marks"] = average_marks
print(df)
```

```python
def grade(avg):
  if avg >= 90:
    return 'A'
  elif avg >= 75:
    return 'B'
  elif avg >= 60:
    return 'C'
  else:
    return 'D'

  df['Grade'] = df['Average Marks'].apply(grade)
  print(df)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df["Name"], df["Average Marks"])
plt.xlabel('Name')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Average marks per subject
df[["Maths", "Scince", "English"]].mean().plot(kind="bar")
plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.show()
```
