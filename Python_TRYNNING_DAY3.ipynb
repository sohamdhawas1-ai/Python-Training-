```python
import numpy as np

# Define your students' marks array (replace with your actual data)
students = np.array([
    [85, 88, 79], # Marks for Student 1
    [55, 62, 68] #Marks for Student 2
])

#Calculate row-wise averages for each student
avg_marks = np.mean(students, axis=1)

#Performance Report
for i, avg in enumerate (avg_marks):
    status = "Pass" if avg > 50 else "Fail"
    print(f"Student (i+1): Avg {avg:.2f} | {status}")
```

```python
import matplotlib.pyplot as plt
# Sample data for demonstration
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [15000, 22000, 18000, 30000, 25000]
# Generating the horizontal bar chart
plt.barh(months, sales, color='grey')
plt.title('Monthly sales (Horizontal)')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

```python
import matplotlib.pyplot as plt
# Scatter Plot (comparing height vs weight)
height = [150, 160, 170, 180]
weight = [50, 60, 65, 75]
plt.scatter(height, weight, color = 'green')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# Pie Chart (Market Share)
brands = ['Brand A', 'Brand B', 'Brand C']
shares = [45, 30, 25]
plt.pie(shares, labels = brands, autopct='%1.1f%%', startangle=90)
plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
# Multiple Plot in one figure
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]
plt.plot(x, y1, label='Square')
plt.plot(x, y2, label='Linear')
plt.legend()
plt.title('Multiple Lines')
plt.show()
```

```python
#subplots
fig, axs=plt.subplots (2,2)

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [15000, 22000, 18000, 30000, 25000]

axs[0,0].plot(x,y1)
axs[0,0].set_title('square')
axs[0,1].bar(months, sales)
axs[0,1].set_title('Sales')
axs[1,0].scatter(height, weight)
axs[1,0].set_title('Scatter')
axs[1,1].pie(shares, labels = brands, autopct='%1.1f%%')
axs[1,1].set_title('Pie')
plt.tight_layout()
plt.show()
```
